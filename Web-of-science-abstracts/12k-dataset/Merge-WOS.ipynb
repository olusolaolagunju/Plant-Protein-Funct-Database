```python
"""
Merge WOS-*.xlsx files into a single Excel file.

Expects filenames like:
    WOS-1-1000.xlsx
    WOS-1001-2000.xlsx
    WOS-2001-3000.xlsx
    ...
    WOS-10001-10972.xlsx

Usage:
    1. Set INPUT_FOLDER below to the folder containing the WOS-*.xlsx files.
    2. Set OUTPUT_FILE to where you want the merged file saved.
    3. Run:  python merge_wos_xlsx.py

Requires: pandas, openpyxl
    pip install pandas openpyxl --break-system-packages   (if needed)
"""

import re
import glob
import os
import pandas as pd

# ---- CONFIG: edit these two paths ----
INPUT_FOLDER = "./WOS"                      # folder containing the WOS-*.xlsx files
OUTPUT_FILE = "./WOS_merged_10972.xlsx"      # merged output file
# ---------------------------------------

# Set to True to drop exact duplicate rows after merging (e.g. if ranges overlap
# or the same record appears in two files). Leave False if you want every row kept.
DROP_DUPLICATES = True

# If your sheets have a unique ID column (e.g. "UT (Unique WOS ID)"), you can
# dedupe on just that column instead of full-row duplicates. Set the column
# name here, or leave as None to dedupe on all columns (only used if
# DROP_DUPLICATES is True).
DEDUPE_COLUMN = "DOI"


def extract_start_number(filepath):
    """Sort files numerically by the first number in the filename, e.g.
    WOS-9001-10000.xlsx -> 9001, so files merge in the right order."""
    name = os.path.basename(filepath)
    match = re.search(r"(\d+)", name)
    return int(match.group(1)) if match else 0


def merge_wos_files(input_folder, output_file):
    files = glob.glob(os.path.join(input_folder, "WOS-*.xlsx"))
    if not files:
        raise FileNotFoundError(f"No WOS-*.xlsx files found in {input_folder}")

    files.sort(key=extract_start_number)

    print(f"Found {len(files)} files to merge:")
    for f in files:
        print(f"  - {os.path.basename(f)}")

    frames = []
    total_rows = 0
    for f in files:
        df = pd.read_excel(f)
        print(f"  {os.path.basename(f)}: {len(df)} rows")
        total_rows += len(df)
        frames.append(df)

    merged = pd.concat(frames, ignore_index=True)
    print(f"\nTotal rows before merge check: {total_rows}")
    print(f"Total rows in merged dataframe: {len(merged)}")

    if DROP_DUPLICATES:
        before = len(merged)
        if DEDUPE_COLUMN and DEDUPE_COLUMN in merged.columns:
            merged = merged.drop_duplicates(subset=[DEDUPE_COLUMN], keep="first")
        else:
            merged = merged.drop_duplicates(keep="first")
        print(f"Dropped {before - len(merged)} duplicate rows")

    merged.to_excel(output_file, index=False, engine="openpyxl")
    print(f"\nSaved merged file to: {output_file}")
    print(f"Final row count: {len(merged)}")


if __name__ == "__main__":
    merge_wos_files(INPUT_FOLDER, OUTPUT_FILE)
```
