```python
"""
Merge phase2_gemini_triples.xlsx and phase2_openai_triples.xlsx into one
combined file for side-by-side review across all 5 models, plus a combined
summary sheet joining cost/latency/failure stats from both provider runs.

Run this after both run_phase2_gemini.py and run_phase2_openai.py have
completed.

Usage:
    python merge_phase2_comparison.py
"""

from pathlib import Path
import pandas as pd

OUTPUT_DIR = Path("model_comparison_output")

triples_files = [
    OUTPUT_DIR / "phase2_gemini_triples.xlsx",
    OUTPUT_DIR / "phase2_openai_triples.xlsx",
]
summary_files = [
    OUTPUT_DIR / "phase2_gemini_summary.xlsx",
    OUTPUT_DIR / "phase2_openai_summary.xlsx",
]

triples = [pd.read_excel(f) for f in triples_files if f.exists()]
missing_triples = [f.name for f in triples_files if not f.exists()]
if missing_triples:
    print(f"Note: missing {missing_triples}, merging what's available.")

if triples:
    combined_triples = pd.concat(triples, ignore_index=True)
    out_path = OUTPUT_DIR / "phase2_all_models_triples.xlsx"
    combined_triples.to_excel(out_path, index=False)
    print(f"Combined triples ({len(combined_triples)} rows, "
          f"{combined_triples['model'].nunique()} models) -> {out_path}")
else:
    print("No triple files found yet, run both provider scripts first.")

summaries = [pd.read_excel(f) for f in summary_files if f.exists()]
missing_summaries = [f.name for f in summary_files if not f.exists()]
if missing_summaries:
    print(f"Note: missing {missing_summaries}, merging what's available.")

if summaries:
    combined_summary = pd.concat(summaries, ignore_index=True)
    out_path = OUTPUT_DIR / "phase2_all_models_summary.xlsx"
    combined_summary.to_excel(out_path, index=False)
    print(f"Combined summary -> {out_path}")
    print("\n" + combined_summary.to_string(index=False))
```
